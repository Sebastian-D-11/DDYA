## 1) Descripción del problema **[1 pts]**
Un vivero esta analizando el crecimiento de la producción de nuevas plantas a partir de un proceso de reproduccion controlada, las plantas producidas al mes tienen el comportamiento de una funcion de recurrencia con casos base; mes 0 - 0 nuevas plantas, mes 1 - 1 nueva planta, a partir del segundo mes el numero de plantas es igual a la suma de las plantas de los dos meses anteriores, el problema consiste en crear un programa que pueda calcular n plantas en el mes n.

## 6) Código funcional **[15 pts]**




```python
def bienvenida():
    print("\t --BIENVENIDO--")
    print("Se realizará el cálculo del mes que desee para ver la producción de plantas en dicho mes.")

def main():
    cierre = 0
    bienvenida()
    
    while cierre != 1:
        num_mes = int(input("\nIngrese el número del mes que desea calcular: "))
        
        while num_mes < 0:
            print("Dato no válido, por favor ingrese un número entero positivo o mayor a cero.")
            num_mes = int(input("\nIngrese el número del mes que desea calcular: "))
            
        p_plantas = [0, 1]
        for i in range(2, num_mes + 1):
            suma_produccion = p_plantas[i-1] + p_plantas[i-2]
            p_plantas.append(suma_produccion)
            
        print("La producción de plantas en el mes solicitado es:", p_plantas[num_mes])
        
        f = input("¿Desea volver a realizar el cálculo de la producción (Si) o (No)?: ")
        
        while f != "si" and f != "no":
            print("Error en la respuesta, por favor responda (si) o (no).")
            f = input("¿Desea volver a realizar el cálculo de la producción (Si) o (No)?: ")
            
        if f == "no":
            cierre = 1
        elif f == "si":
            cierre = 0
            
    print("Muchas gracias por su tiempo, feliz día.")

main()
```
